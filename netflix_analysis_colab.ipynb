```python
# Fill missing values with NA
df_cleaned = df.fillna("NA")
```

```python
# Question 7: How many Movies and TV Shows are there?
type_count = df_cleaned['type'].value_counts()

print(type_count)
```

```python
type_count.plot(kind='bar')

plt.title("Movies vs TV Shows")
plt.xlabel("Type")
plt.ylabel("Number of Titles")
plt.show()
```

```python
# Question 8: Which year had the most content added to Netflix?

df_cleaned['date_added'] = pd.to_datetime(
    df_cleaned['date_added'],
    errors='coerce'
)

content_added_year = df_cleaned['date_added'].dt.year.value_counts().sort_index()

print(content_added_year)
```

```python
content_added_year.plot(kind='line', marker='o')

plt.title("Netflix Content Added by Year")
plt.xlabel("Year")
plt.ylabel("Number of Titles")
plt.show()
```

```python
# Question 9: Which countries have the highest number of Netflix titles?

country_df = df_cleaned[['show_id', 'title', 'country']].copy()

country_df['country'] = country_df['country'].str.split(', ')

country_df = country_df.explode('country')

country_count = country_df['country'].value_counts()

print(country_count.head(10))
```

```python
country_count.head(10).sort_values().plot(kind='barh')

plt.title("Top 10 Countries by Number of Netflix Titles")
plt.xlabel("Number of Titles")
plt.ylabel("Country")
plt.show()
```

```python
# Question 10: What are the most common content ratings?

rating_count = df_cleaned['rating'].value_counts()

print(rating_count)
```

```python
rating_count.plot(kind='bar')

plt.title("Most Common Content Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
```

```python
# Question 11: What are the most popular genres?

genre_df = df_cleaned[['show_id', 'title', 'listed_in']].copy()

genre_df['listed_in'] = genre_df['listed_in'].str.split(', ')

genre_df = genre_df.explode('listed_in')

genre_count = genre_df['listed_in'].value_counts()

print(genre_count.head(10))
```

```python
genre_count.head(10).sort_values().plot(kind='barh')

plt.title("Top 10 Most Popular Genres")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")
plt.show()
```

```python
# Question 12: Which type of content dominates Netflix?

type_count = df_cleaned['type'].value_counts()

print(type_count)
print("Dominant Type:", type_count.idxmax())
```

```python
# Question 13: Which year had the highest number of Movies released?

movies_df = df_cleaned[df_cleaned['type'] == 'Movie']

movie_release_year = movies_df['release_year'].value_counts().sort_index()

print(movie_release_year)

print(
    "Year with highest number of Movies:",
    movie_release_year.idxmax()
)
```

```python
movie_release_year.plot(kind='line', marker='o')

plt.title("Movies Released by Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Movies")
plt.show()
```

```python
# Question 14: Which year had the highest number of TV Shows released?

tv_df = df_cleaned[df_cleaned['type'] == 'TV Show']

tv_release_year = tv_df['release_year'].value_counts().sort_index()

print(tv_release_year)

print(
    "Year with highest number of TV Shows:",
    tv_release_year.idxmax()
)
```

```python
tv_release_year.plot(kind='line', marker='o')

plt.title("TV Shows Released by Year")
plt.xlabel("Release Year")
plt.ylabel("Number of TV Shows")
plt.show()
```

```python
# Question 15: What is the average release year?

average_release_year = df_cleaned['release_year'].mean()

print("Average Release Year:", average_release_year)
```

```python
# Question 16: What is the oldest content on Netflix?

oldest_year = df_cleaned['release_year'].min()

oldest_content = df_cleaned[
    df_cleaned['release_year'] == oldest_year
]

print("Oldest Release Year:", oldest_year)
print(oldest_content[['title', 'type', 'release_year']])
```

```python
# Question 17: What is the latest release year available?

latest_year = df_cleaned['release_year'].max()

latest_content = df_cleaned[
    df_cleaned['release_year'] == latest_year
]

print("Latest Release Year:", latest_year)
print(latest_content[['title', 'type', 'release_year']])
```

```python
# Question 18: Which directors have the most content?

director_df = df_cleaned.copy()

director_df['director'] = director_df['director'].str.split(', ')

director_df = director_df.explode('director')

director_count = director_df['director'].value_counts()

print(director_count.head(10))
```

```python
director_count.head(10).sort_values().plot(kind='barh')

plt.title("Top 10 Directors by Number of Titles")
plt.xlabel("Number of Titles")
plt.ylabel("Director")
plt.show()
```

```python
# Question 19: How has Netflix content changed over the years?

content_by_year = (
    df_cleaned
    .groupby(['release_year', 'type'])
    .size()
    .unstack(fill_value=0)
)

print(content_by_year)
```

```python
content_by_year.plot(kind='line', marker='o')

plt.title("Netflix Content Trend Over the Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.show()
```

```python
# Question 20: Which countries produced the most Movies?

movie_country_df = movies_df[['show_id', 'title', 'country']].copy()

movie_country_df['country'] = movie_country_df['country'].str.split(', ')

movie_country_df = movie_country_df.explode('country')

top_movie_countries = movie_country_df['country'].value_counts()

print(top_movie_countries.head(10))
```

```python
top_movie_countries.head(10).sort_values().plot(kind='barh')

plt.title("Top 10 Countries Producing Movies")
plt.xlabel("Number of Movies")
plt.ylabel("Country")
plt.show()
```
